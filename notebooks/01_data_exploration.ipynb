```python
test_records = []

with open(DATA_FILE, "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i >= 100:
            break

        test_records.append(json.loads(line))

print(f"Loaded {len(test_records)} records")
```

```python
OBSERVATIONS = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "observations"
)

print(OBSERVATIONS)
```

```python
con.execute(f"""
    SELECT COUNT(*) AS total_records
    FROM '{OBSERVATIONS}/*.parquet'
""").df()
```

```python
con.execute(f"""
    SELECT
        COUNT(*) AS observations,
        COUNT(DISTINCT organization) AS organizations,
        COUNT(DISTINCT ip) AS unique_ips,
        COUNT(DISTINCT port) AS unique_ports,
        COUNT(DISTINCT product) AS unique_products,
        COUNT(DISTINCT country_code) AS countries
    FROM '{OBSERVATIONS}/*.parquet'
""").df()
```

```python
# Finding the biggest organizations
con.execute(f"""
    SELECT
        organization,
        COUNT(*) AS observation_count,
        COUNT(DISTINCT ip) AS unique_ips,
        COUNT(DISTINCT product) AS unique_products
    FROM '{OBSERVATIONS}/*.parquet'
    WHERE organization IS NOT NULL
    GROUP BY organization
    ORDER BY observation_count DESC
    LIMIT 20
""").df()
```

```python
# Check Countries
con.execute(f"""
    SELECT
        country_code,
        country_name,
        COUNT(*) AS observations,
        COUNT(DISTINCT organization) AS organizations
    FROM '{OBSERVATIONS}/*.parquet'
    GROUP BY country_code, country_name
    ORDER BY observations DESC
    LIMIT 20
""").df()
```

```python
# Check products
con.execute(f"""
    SELECT
        product,
        COUNT(*) AS observations,
        COUNT(DISTINCT organization) AS organizations
    FROM '{OBSERVATIONS}/*.parquet'
    WHERE product IS NOT NULL
      AND TRIM(product) != ''
    GROUP BY product
    ORDER BY observations DESC
    LIMIT 30
""").df()
```

```python
# Check exposed ports
con.execute(f"""
    SELECT
        port,
        COUNT(*) AS observations,
        COUNT(DISTINCT organization) AS organizations
    FROM '{OBSERVATIONS}/*.parquet'
    WHERE port IS NOT NULL
    GROUP BY port
    ORDER BY observations DESC
    LIMIT 30
""").df()
```

```python
# investigating security-relevant services
#SSH
con.execute(f"""
    SELECT
        organization,
        product,
        port,
        COUNT(*) AS observations
    FROM '{OBSERVATIONS}/*.parquet'
    WHERE
        LOWER(COALESCE(product, '')) LIKE '%ssh%'
        OR port = 22
    GROUP BY organization, product, port
    ORDER BY observations DESC
    LIMIT 30
""").df()
```

```python
#HTTP
con.execute(f"""
    SELECT
        organization,
        product,
        port,
        COUNT(*) AS observations
    FROM '{OBSERVATIONS}/*.parquet'
    WHERE
        port IN (80, 443)
    GROUP BY organization, product, port
    ORDER BY observations DESC
    LIMIT 30
""").df()
```
